```python
def get_data(name):


    url = "https://catalogo.uc.cl/index.php?tmpl=component&option=com_catalogo&view=programa&sigla="+name

    response = requests.get(url)
    response.raise_for_status()

    soup = BeautifulSoup(response.text, "html.parser")

    # Obtener el contenido del <pre>
    pre = soup.find("pre")

    if pre:
        texto = pre.get_text("\n", strip=True)

        # Extraer traducción
        traduccion = re.search(
            r"TRADUCCION:\s*(.+)",
            texto
        )

        # Extraer sigla
        sigla = re.search(
            r"SIGLA:\s*(.+)",
            texto
        )

        # Extraer descripción
        descripcion = re.search(
            r"I\.\s*DESCRIPCION\s*(.*?)\s*II\.\s*OBJETIVOS",
            texto,
            re.DOTALL
        )

        datos = {
            "sigla": sigla.group(1).strip() if sigla else None,
            "traduccion": traduccion.group(1).strip() if traduccion else None,
            "descripcion": descripcion.group(1).strip() if descripcion else None,
    }
        
    return datos
```
